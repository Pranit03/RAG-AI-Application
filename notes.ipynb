```python
from pypdf import PdfReader
from groq import Groq
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# Read PDF

reader = PdfReader("AI Engineer Notes 529 Pages.pdf")
```

```python
print(reader.pages)
```

```python
text = " "
for page in reader.pages:
    text += page.extract_text()

print(text)
```

```python
client = Groq(api_key= os.getenv("GROQ_API_KEY"))
```

```python
question =input("Ask Your Quation: ")
```

```python
question =input("Ask Your Quation: ")

pdf_content = text[:5000]  # first 5000 characters

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "system",
            "content": "Answer only from the provided PDF content."
        },
        {
            "role": "user",
            "content": f"PDF Content:\n{pdf_content}\n\nQuestion: {question}"
        }
    ]
)

print(response.choices[0].message.content)
```
